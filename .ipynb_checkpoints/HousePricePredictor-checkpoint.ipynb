## Correlation

```python
plt.figure(figsize=(32,16))
sns.heatmap(df_train.corr(numeric_only=True), annot=True, cmap='coolwarm')
```

## Scatter Plots

```python
plt.figure(figsize=(16,9))

plt.subplot(2, 2, 1)
sns.scatterplot(data=df_train, x="GrLivArea", y="LogSalePrice")

plt.subplot(2, 2, 2)
sns.scatterplot(data=df_train, x="TotalBsmtSF", y="LogSalePrice")

plt.subplot(2, 2, 3)
sns.scatterplot(data=df_train, x="GarageArea", y="LogSalePrice")

plt.subplot(2, 2, 4)
sns.scatterplot(data=df_train, x="1stFlrSF", y="LogSalePrice")

plt.show()
```
